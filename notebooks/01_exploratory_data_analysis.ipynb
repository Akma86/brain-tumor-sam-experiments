# Brain Tumor MRI Classification — Exploratory Data Analysis (EDA)

This notebook explores the **Brain Tumor MRI Dataset** sourced from **CSCR Hospital** and **Epic Health Care** (Chittagong, Bangladesh).

### Target Classes:
1. `glioma` (Malignant brain tumor originating from glial tissue)
2. `meningioma` (Primary tumor originating in brain and spinal cord membranes)
3. `notumor` (Healthy control scans)
4. `pituitary` (Tumor developing in the pituitary gland)

```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from PIL import Image
```

## 1. Class Distribution Analysis

```python
data_dir = Path("../data")
records = []

for split in ["train", "test"]:
    split_dir = data_dir / split
    if split_dir.exists():
        for class_dir in sorted(split_dir.iterdir()):
            if class_dir.is_dir():
                count = len(list(class_dir.glob("*.jpg")))
                records.append({"Split": split.capitalize(), "Class": class_dir.name, "Count": count})

df = pd.DataFrame(records)
df_pivot = df.pivot(index="Class", columns="Split", values="Count")
df_pivot["Total"] = df_pivot["Train"] + df_pivot["Test"]
df_pivot["Percentage (%)"] = (df_pivot["Total"] / df_pivot["Total"].sum() * 100).round(2)
display(df_pivot)
```

## 2. Visualize Sample MRI Scans per Category

```python
classes = ["glioma", "meningioma", "notumor", "pituitary"]
fig, axes = plt.subplots(len(classes), 4, figsize=(12, 12))

for row_idx, cls in enumerate(classes):
    cls_path = data_dir / "train" / cls
    sample_images = sorted(list(cls_path.glob("*.jpg")))[:4]
    for col_idx, img_path in enumerate(sample_images):
        ax = axes[row_idx, col_idx]
        img = Image.open(img_path)
        ax.imshow(img, cmap="bone")
        ax.axis("off")
        if col_idx == 0:
            ax.set_title(cls.capitalize(), fontsize=14, weight="bold", loc="left")

plt.suptitle("Sample MRI Scans Across Diagnostic Categories", fontsize=16, weight="bold", y=0.92)
plt.tight_layout()
plt.show()
```

## 3. Pixel Intensity Distribution
Examine brightness distribution and contrast profiles across MRI samples.

```python
plt.figure(figsize=(10, 5))
for cls in classes:
    sample_path = next((data_dir / "train" / cls).glob("*.jpg"))
    img_arr = np.array(Image.open(sample_path).convert("L"))
    plt.hist(img_arr.ravel(), bins=50, alpha=0.5, label=cls.capitalize(), density=True)

plt.title("Grayscale Pixel Intensity Distribution Across Classes")
plt.xlabel("Pixel Intensity (0-255)")
plt.ylabel("Density")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
